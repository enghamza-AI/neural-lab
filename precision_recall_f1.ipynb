```python
import numpy as np
```

```python
# Creating toy imbalanced dataset

y_true = np.array([0] * 995 + [1] * 5)
y_pred = np.array([0] * 1000)
```

```python
# computing confusion matrix manually

TP = np.sum((y_pred == 1) & (y_true == 1))
FP = np.sum((y_pred == 1) & (y_true == 0))
FN = np.sum((y_pred == 0) & (y_true == 1))
TN = np.sum((y_pred == 0) & (y_true == 0))

print(f"TP: {TP}, FP: {FP}, FN: {FN}, TN: {TN}")
```

```python
# compute precision and recall

precision = TP / (TP + FP) if (TP + FP) > 0 else 0
recall = TP / (TP + FN) if (TP + FN) > 0 else 0

print(f"Precision: {precision:.2f}, Recall: {recall:.2f}")
```

```python
# Computing Accuracy

accuracy = (TP + TN) / (TP + TN + FP + FN)
print(f"Accuracy: {accuracy:.2f}")
```

```python
# verifying with Sklearn

from sklearn.metrics import confusion_matrix, precision_score, recall_score, accuracy_score

# Confusion matrix
cm = confusion_matrix(y_true, y_pred)
print("Confusion Matrix (sklearn):\n", cm)

# Precision, recall, accuracy
sklearn_precision = precision_score(y_true, y_pred)
sklearn_recall = recall_score(y_true, y_pred)
sklearn_accuracy = accuracy_score(y_true, y_pred)

print(f"Precision (sklearn): {sklearn_precision:.2f}")
print(f"Recall (sklearn): {sklearn_recall:.2f}")
print(f"Accuracy (sklearn): {sklearn_accuracy:.2f}")
```

```python
# Lesson

# Why 99% accuracy is useless:
# In this imbalanced dataset (995 negatives, 5 positives),
# a model that always predicts 0 will be correct 99.5% of the time.
# But it fails to detect ANY positives (recall = 0).
# This is why precision, recall, and F1-score are more important for imbalanced data.
```

```python
# Trying a better model

# New predictions: 4 true positives, 1 false negative, 0 false positives
y_pred_improved = np.array([0] * 995 + [1] * 4 + [0] * 1)

# Recompute metrics
TP = np.sum((y_pred_improved == 1) & (y_true == 1))
FP = np.sum((y_pred_improved == 1) & (y_true == 0))
FN = np.sum((y_pred_improved == 0) & (y_true == 1))
TN = np.sum((y_pred_improved == 0) & (y_true == 0))

precision = TP / (TP + FP) if (TP + FP) > 0 else 0
recall = TP / (TP + FN) if (TP + FN) > 0 else 0
accuracy = (TP + TN) / (TP + TN + FP + FN)

print(f"TP: {TP}, FP: {FP}, FN: {FN}, TN: {TN}")
print(f"Precision: {precision:.2f}, Recall: {recall:.2f}, Accuracy: {accuracy:.2f}")
```

```python
# SUMMARY OF WHAT I HAVE LEARNED 
'''
Confusion Matrix: TP, FP, FN, TN are the building blocks.
Precision/Recall: Focus on the positive class, not just overall accuracy.
Imbalanced Data: Accuracy can be misleading. Always check precision and recall.
sklearn Verification: Your manual calculations should match sklearn’s output.

'''
```
